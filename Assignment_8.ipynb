**Packages needed to be imported**

```python
import numpy as np
import pandas as pd
from tensorflow.keras import layers,models
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.preprocessing.sequence import pad_sequences
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import imdb
from sklearn.model_selection import train_test_split
```

**Loading the data**

```python
max_feature=1000
max_len=100
```

```python
(X_train,y_train),(X_test,y_test)=imdb.load_data(num_words=max_feature)
```

```python
X_train=pad_sequences(X_train,maxlen=max_len)
X_test=pad_sequences(X_test,maxlen=max_len)
```

**Model Building**

```python
model_rnn=models.Sequential()
model_rnn.add(layers.Embedding(max_feature,32))
model_rnn.add(layers.SimpleRNN(64))
model_rnn.add(layers.Dense(1,activation='sigmoid'))
```

```python
model_rnn.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history=model_rnn.fit(X_train,y_train,epochs=50,validation_split=0.2,batch_size=64)
```

```python
loss,accuracy=model_rnn.evaluate(X_test,y_test)
print(f"loss {loss}")
print(f'accuracy {accuracy*100}')
```

```python
predication_rnn=model_rnn.predict(X_test[:5])
for i,prediction in enumerate(predication_rnn):
  sentiment= 'Positive' if prediction>0.5 else 'Negative'
  print(f'Review {i+1} Predcition is {sentiment} ,actual prediction is {"Positive" if y_test[i]==1 else "Negative"}')
```
